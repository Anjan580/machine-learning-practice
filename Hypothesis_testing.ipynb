```python
from scipy.stats import binom
```

```python
n = 10
x = 2
p = 0.5
alpha = 0.05
```

```python
# p_value = binom(observed, total , probability)
p_value = binom.cdf(x,n,p)
print(p_value)
```

```python
# decision
if p_value < alpha :
    print('do not accept H0')
else:
    print('Accept H0')
```
